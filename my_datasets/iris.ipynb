## what鸢尾花数据
鸢尾花数据集是机器学习领域中一个非常著名且经典的数据集，由英国统计学家和生物学家罗纳德·费舍尔（Ronald Fisher）在1936年首次提出。

鸢尾花数据集的特点：

这个数据集包含150个鸢尾花样本，每个样本有四个特征，分别是花萼（sepal）长度、花萼宽度、花瓣（petal）长度和花瓣宽度，以及它们所属的鸢尾花品种。
150个鸢尾花样本包括三种不同的品种：山鸢尾（Setosa）、变色鸢尾（Versicolor）、维吉尼亚鸢尾（Virginica），每种品种各有50个样本。

```python
# 双变量分析：散点图矩阵
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# 加载数据
iris = load_iris()

# 构建完整的 DataFrame（含特征 + 物种名称）
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)

# 设置中文字体（可选）
plt.rcParams['font.sans-serif'] = ['PingFang SC', 'Hiragino Sans GB', 'Arial Unicode MS', 'Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

# 绘制 pairplot
plt.figure(figsize=(8, 6))
g = sns.pairplot(
    df,  # ← 关键：传入的是 df，不是 iris
    hue='species',
    diag_kind='hist',
    palette={'setosa': 'red', 'versicolor': 'green', 'virginica': 'blue'},
    plot_kws={'alpha': 0.7, 's': 50}
)

g.figure.suptitle('鸢尾花数据集：散点图矩阵', y=1.02, fontsize=16, fontweight='bold')
plt.show()
```

```python
# 双变量分析：联合分布
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# 加载数据
iris = load_iris()

# 创建一个 DataFrame 并添加列名
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target_names[iris.target]

# 绘制 jointplot
g = sns.jointplot(
    data=df,
    x='petal length (cm)',  # 注意这里的列名应该和 df 中的一致
    y='petal width (cm)',   # 同上
    hue='species',
    kind='scatter',
    height=6,
    ratio=5,
    s=100,  # 调整点大小以便更好地展示
    alpha=0.7,
)

g.figure.suptitle('花瓣长度与宽度的联合分布', y=1.02, fontsize=14, fontweight='bold')
plt.show()
```

```python
# 多变量分析：3D散点图
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# 加载鸢尾花数据集
iris = load_iris()

# 创建包含特征和物种名称的完整 DataFrame
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)

# 设置中文字体和图表样式
plt.rcParams['font.sans-serif'] = ['PingFang SC', 'Hiragino Sans GB', 'Arial Unicode MS', 'Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

# 创建3D散点图
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

colors = {'setosa': 'red', 'versicolor': 'green', 'virginica': 'blue'}

for species in df['species'].unique():
    species_data = df[df['species'] == species]
    ax.scatter(
        species_data['petal length (cm)'],
        species_data['petal width (cm)'],
        species_data['sepal length (cm)'],
        c=colors[species],
        label=species,
        s=50,
        alpha=0.7
    )

ax.set_xlabel('花瓣长度 (cm)')
ax.set_ylabel('花瓣宽度 (cm)')
ax.set_zlabel('萼片长度 (cm)')
ax.set_title('鸢尾花3D散点图', fontsize=14, fontweight='bold')
ax.legend()
plt.show()
```

## 4. 数据清洗
异常值处理，缺失值处理，重复值处理

鸢尾花数据比较干净，主动添加脏数据，然后处理掉。

添加脏数据

```python
import pandas as pd
from sklearn.datasets import load_iris

# 加载鸢尾花数据
iris = load_iris()
data = pd.DataFrame(data=iris.data, columns=iris.feature_names)
data['species'] = iris.target_names[iris.target]
print("缺失值检查 ")
missing_values = data.isnull().sum()
print(missing_values)
print("检查重复值")
duplicates = data.duplicated().sum()
print(f"重复行数: {duplicates}")

# 添加异常值150~152
outliers = pd.DataFrame([
    [59, 3.0, 5.1, 1.8, 'unknown'],
    [5.9, 30, 5.1, 1.8, 'unknown'],
    [5.9, 3.0, 51, 1.8, 'unknown']
], columns=data.columns)

data = pd.concat([data, outliers], ignore_index=True)

# 添加空值153~155
empyties = pd.DataFrame([
    [5.9, None, None, 1.8, 'unknown'],
    [None, 3.0, 5.1, 1.8, 'unknown'],
    [5.9, None, 5.1, 1.8, 'unknown']
], columns=data.columns)
data = pd.concat([data, empyties], ignore_index=True)

# 添加重复值156~160
for i in range(156,160):
    data.loc[i] = [5.9, 3.0, 5.1, 1.8 , "unknown"]

# 查看数据
data.tail(12)
print("#"*26,"修改之后","#"*26)
print("缺失值检查 ")
missing_values = data.isnull().sum()
print(missing_values)

print("检查重复值")
duplicates = data.duplicated().sum()
print(f"重复行数: {duplicates}")
```

```python
# 处理脏数据
# 删除缺失值，空值：inplace = True，修改源数据
data.dropna(inplace = True)
# 重置索引
data.reset_index(drop=True, inplace=True)
# data.fillna(value)	将缺失值替换为指定的值
data.tail(12)
# data.info()

# 删除重复数据
data.drop_duplicates(inplace = True)
# 重置索引
data.reset_index(drop=True, inplace=True)
data.tail(12)
# data.info()

# 处理异常值
# 删除特定列中的异常值
data = data[(data['sepal length (cm)'] <= 10) & (data['sepal width (cm)'] <= 10)]
data = data[(data['petal length (cm)'] <= 10) & (data['petal width (cm)'] <= 10)]
# 重置索引
data.reset_index(drop=True, inplace=True)
# 查看处理后的数据
data.tail(10)
# data.info()
```

## 5. 特征处理
标准化：通过将数据转换为均值为0、标准差为1的分布，使数据集中在0附近



```python
from sklearn.preprocessing import StandardScaler
import pandas as pd
from sklearn.datasets import load_iris

# 加载鸢尾花数据
iris = load_iris()
data = pd.DataFrame(data=iris.data, columns=iris.feature_names)
data['species'] = iris.target_names[iris.target]

X = data.iloc[:, 0:4]
y = data.iloc[:, 4]
print(X[:2])
scaler = StandardScaler().fit(X)
X = scaler.transform(X)
print(X[0:2,:])

# 注意：fit()函数只能作用在训练集上，如果希望对测试集变换，只要用训练集上fit好的转换器去transform即可。不能在测试集上fit再transform，否则训练集和测试集的变换规则会不一致。
```

## 6. 数据拆分
